```python
# ==========================================================
# IMPORTS
# ==========================================================

import os
import numpy as np
import pandas as pd
```

```python
# ==========================================================
# PATHS
# ==========================================================

MASTER_PIPELINE = "/content/drive/MyDrive/Exoplanet_Dataset/08_Master_Pipeline/Master_Pipeline.csv"

PHYSICS_V2 = "/content/drive/MyDrive/Exoplanet_Dataset/10_Physics_Features_V2/Physics_Features_V2.csv"


OUTPUT_FOLDER = "/content/drive/MyDrive/Exoplanet_Dataset/11_Training_Dataset_V2"


os.makedirs(

    OUTPUT_FOLDER,

    exist_ok=True

)
```

```python
# ==========================================================
# LOAD DATASETS
# ==========================================================

master = pd.read_csv(

    MASTER_PIPELINE

)

physics = pd.read_csv(

    PHYSICS_V2

)

print("="*60)
print("MASTER PIPELINE")
print("="*60)

print(master.shape)

print()

print("="*60)
print("PHYSICS FEATURES V2")
print("="*60)

print(physics.shape)
```

```python
print(master.columns.tolist())
```

```python
print("=" * 70)
print("MASTER PIPELINE")
print("=" * 70)

print()

print("Rows    :", len(master))
print("Columns :", len(master))

print()

print(master.columns.tolist())
```

```python
# ==========================================================
# LOAD TRANSIT FEATURES
# ==========================================================

TRANSIT_FEATURES = "/content/drive/MyDrive/Exoplanet_Dataset/09_Transit_Features/Transit_Features.csv"

transit_features = pd.read_csv(

    TRANSIT_FEATURES

)

print("=" * 70)
print("TRANSIT FEATURES")
print("=" * 70)

print()

print("Rows    :", len(transit_features))
print("Columns :", len(transit_features))

print()

display(transit_features.head())
```

```python
print("=" * 70)
print("TRANSIT FEATURES")
print("=" * 70)

print()

print("Rows    :", len(transit_features))
print("Columns :", len(transit_features))

print()

print(transit_features.columns.tolist())
```

```python
# ==========================================================
# BUILD TRAINING DATASET V2
# ==========================================================

training_dataset_v2 = (

    master

    .merge(

        transit_features,

        on="TIC_ID",

        how="inner"

    )

    .merge(

        physics,

        on="TIC_ID",

        how="inner"

    )

)

print("=" * 70)
print("TRAINING DATASET V2")
print("=" * 70)

print()

print("Rows    :", len(training_dataset_v2))
print("Columns :", len(training_dataset_v2.columns))

print()

display(training_dataset_v2.head())
```

```python
OUTPUT_FILE = "/content/drive/MyDrive/Exoplanet_Dataset/11_Training_Dataset_V2/Training_Dataset_V2.csv"

training_dataset_v2.to_csv(

    OUTPUT_FILE,

    index=False

)

print("Saved Successfully!")
print(OUTPUT_FILE)
```
